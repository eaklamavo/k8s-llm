# Lesson 3.2 - Install ArgoCD With Flux HelmRelease

**Level: Intermediate**

## Learning Objectives

**Project file(s) produced:** `bootstrap/argocd-helmrelease.yaml`, `bootstrap/kustomization.yaml`

## Concepts
A `HelmRelease` tells Flux what chart/version to install and how to configure values.
Kustomize then bundles the namespace, source, and release manifests into one reconciliation unit.

```python
%%writefile bootstrap/argocd-helmrelease.yaml
apiVersion: helm.toolkit.fluxcd.io/v2
kind: HelmRelease
metadata:
  name: argocd
  namespace: flux-system
spec:
  interval: 30m
  targetNamespace: argocd
  chart:
    spec:
      chart: argo-cd
      version: ">=7.0.0 <8.0.0"
      sourceRef:
        kind: HelmRepository
        name: argo-helm
        namespace: flux-system
  install:
    remediation:
      retries: 3
    createNamespace: true
  upgrade:
    remediation:
      retries: 3
  values:
    configs:
      params:
        server.insecure: true
    dex:
      enabled: false
    notifications:
      enabled: false
    server:
      service:
        type: LoadBalancer
```

```python
%%writefile bootstrap/kustomization.yaml
apiVersion: kustomize.config.k8s.io/v1beta1
kind: Kustomization

resources:
  - namespaces.yaml
  - argocd-helmrepository.yaml
  - argocd-helmrelease.yaml
```
