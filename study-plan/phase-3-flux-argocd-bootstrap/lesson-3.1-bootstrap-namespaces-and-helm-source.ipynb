**Level: Beginner**

## Learning Objectives
- Create the ArgoCD namespace manifest.
- Register the Argo Helm chart repository in Flux.
- Understand Kustomize composition in bootstrap path.

**Project file(s) produced:** `bootstrap/namespaces.yaml`, `bootstrap/argocd-helmrepository.yaml`

## Concepts
Flux uses source objects (like HelmRepository) and reconciliation objects (like HelmRelease).
Namespace creation is explicit for clarity and GitOps drift control.

```python
%%writefile bootstrap/namespaces.yaml
apiVersion: v1
kind: Namespace
metadata:
  name: argocd
  labels:
    app.kubernetes.io/part-of: argocd
```

```python
%%writefile bootstrap/argocd-helmrepository.yaml
apiVersion: source.toolkit.fluxcd.io/v1
kind: HelmRepository
metadata:
  name: argo-helm
  namespace: flux-system
spec:
  interval: 1h
  url: https://argoproj.github.io/argo-helm
```
