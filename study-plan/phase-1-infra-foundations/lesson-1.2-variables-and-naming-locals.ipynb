## Learning Objectives

**Project file(s) produced:** `terraform/foundation/variables.tf`, `terraform/foundation/locals.tf`

## Concepts
**Variables** let you customize deployments without editing resource blocks.

**Locals** are computed values used to avoid repeating long naming expressions.

```python
%%writefile terraform/foundation/variables.tf
variable "project" {
  type    = string
  default = "cnandk8s"
}

variable "landscape" {
  type    = string
  default = "sbox"
}

variable "region_code" {
  type    = string
  default = "uks"
}

variable "sequence" {
  type    = string
  default = "01"
}

variable "location" {
  type    = string
  default = "uksouth"
}

variable "resource_group_name" {
  type    = string
  default = null
}

variable "vnet_address_space" {
  type    = list(string)
  default = ["10.20.0.0/16"]
}

variable "subnets" {
  type = map(object({
    purpose          = string
    address_prefixes = list(string)
  }))
  default = {
    aks = {
      purpose          = "k8snodes"
      address_prefixes = ["10.20.0.0/22"]
    }
  }
}
```

```python
%%writefile terraform/foundation/locals.tf
locals {
  name_suffix = "${var.project}-${var.landscape}-${var.region_code}-${var.sequence}"

  names = {
    resource_group = "rg-${local.name_suffix}"
    vnet           = "vnet-${local.name_suffix}"
    aks            = "aks-${local.name_suffix}"
    flux_extension = "flux-${local.name_suffix}"
    flux_config    = "fc-${local.name_suffix}"
  }

  subnet_names = {
    for key, subnet in var.subnets :
    key => "snet-${var.project}-${subnet.purpose}-${var.landscape}-${var.region_code}-${var.sequence}"
  }
}
```
