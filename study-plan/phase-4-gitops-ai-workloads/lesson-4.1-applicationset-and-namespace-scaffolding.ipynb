# Lesson 4.1 - ApplicationSet And Namespace Scaffolding

**Level: Intermediate**

## Learning Objectives
- Define the `ai-demos` namespace.
- Create an ApplicationSet that auto-discovers `gitops/*` folders.
- Understand Git directory generator behavior.

**Project file(s) produced:** `gitops/ai-demos-namespace/namespace.yaml`, `gitops/ai-demos-namespace/kustomization.yaml`, `gitops/applicationset.yaml`, `gitops/kustomization.yaml`

```python
%%writefile gitops/ai-demos-namespace/namespace.yaml
apiVersion: v1
kind: Namespace
metadata:
  name: ai-demos
  labels:
    app.kubernetes.io/part-of: ai-demos
```

```python
%%writefile gitops/applicationset.yaml
apiVersion: argoproj.io/v1alpha1
kind: ApplicationSet
metadata:
  name: ai-demos
  namespace: argocd
spec:
  generators:
    - git:
        repoURL: git@github.com:sl31pn1r/aionk8s.git
        revision: master
        directories:
          - path: gitops/*
  template:
    metadata:
      name: "{{path.basename}}"
      labels:
        app.kubernetes.io/managed-by: argocd
        demo: ai-platform
    spec:
      project: default
      source:
        repoURL: git@github.com:sl31pn1r/aionk8s.git
        targetRevision: master
        path: "{{path}}"
      destination:
        server: https://kubernetes.default.svc
        namespace: ai-demos
      syncPolicy:
        automated:
          prune: true
          selfHeal: true
        syncOptions:
          - CreateNamespace=true
```

```python
%%writefile gitops/kustomization.yaml
apiVersion: kustomize.config.k8s.io/v1beta1
kind: Kustomization

resources:
  - applicationset.yaml
```

```python
%%writefile gitops/ai-demos-namespace/kustomization.yaml
apiVersion: kustomize.config.k8s.io/v1beta1
kind: Kustomization

resources:
  - namespace.yaml
```
